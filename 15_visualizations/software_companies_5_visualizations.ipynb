```python

from pathlib import Path
import pandas as pd
```

```python
import matplotlib.pyplot as plt
import seaborn as sns
import plotly.express as px
sns.set_theme(style="whitegrid")
```

## Load the dataset

```python
filename = "software_companies_dataset_v2 (3)(1).csv"
df = pd.read_csv(filename)
df.head()
```

# Matplotlib Visualizations

### 1. Bar chart — companies by risk rating

```python
df["Risk_Rating"].value_counts().plot(kind="bar", color="skyblue")
plt.title("Companies by Risk Rating")
plt.xlabel("Risk Rating")
plt.ylabel("Number of Companies")
plt.show()
```

### 2. Histogram — employee distribution

```python
plt.hist(df["Employees"].dropna(), bins=15, color="orange", edgecolor="black")
plt.title("Employee Distribution")
plt.xlabel("Number of Employees")
plt.ylabel("Frequency")
plt.show()
```

### 3. Scatter plot — employees versus annual revenue

```python
plt.scatter(df["Employees"], df["Annual_Revenue"], color="green", alpha=0.5)
plt.title("Employees vs Annual Revenue")
plt.xlabel("Employees")
plt.ylabel("Annual Revenue")
plt.show()
```

### 4. Line chart — employees in the first 20 companies

```python
plt.plot(df["Company_Name"].head(20), df["Employees"].head(20), marker="o")
plt.title("Employees in First 20 Companies")
plt.xlabel("Company")
plt.ylabel("Employees")
plt.xticks(rotation=90)
plt.show()
```

### 5. Pie chart — ownership-type distribution

```python
ownership_counts = df["Ownership_Type"].value_counts()
plt.pie(ownership_counts, labels=ownership_counts.index, autopct="%1.1f%%")
plt.title("Ownership Type Distribution")
plt.show()
```

# Seaborn Visualizations

### 6. Count plot — companies by primary cloud

```python
sns.countplot(data=df, x="Primary_Cloud")
plt.title("Companies by Primary Cloud")
plt.xlabel("Primary Cloud")
plt.ylabel("Number of Companies")
plt.xticks(rotation=45)
plt.show()
```

### 7. Histogram — profit-margin distribution

```python
sns.histplot(data=df, x="Profit_Margin", bins=15, kde=True)
plt.title("Profit Margin Distribution")
plt.xlabel("Profit Margin")
plt.show()
```

### 8. Box plot — profit margin by risk rating

```python
sns.boxplot(data=df, x="Risk_Rating", y="Profit_Margin")
plt.title("Profit Margin by Risk Rating")
plt.xlabel("Risk Rating")
plt.ylabel("Profit Margin")
plt.show()
```

### 9. Scatter plot — AI adoption versus cloud adoption

```python
sns.scatterplot(data=df, x="Adoption_Rate_AI", y="Adoption_Rate_Cloud")
plt.title("AI Adoption vs Cloud Adoption")
plt.xlabel("AI Adoption Rate")
plt.ylabel("Cloud Adoption Rate")
plt.show()
```

### 10. Violin plot — satisfaction by ownership type

```python
sns.violinplot(data=df, x="Ownership_Type", y="Employee_Satisfaction")
plt.title("Employee Satisfaction by Ownership Type")
plt.xlabel("Ownership Type")
plt.ylabel("Employee Satisfaction")
plt.xticks(rotation=30)
plt.show()
```

# Plotly Visualizations

### 11. Interactive bar chart — companies by country

```python
country_counts = df["Country"].value_counts().reset_index()
country_counts.columns = ["Country", "Companies"]
fig = px.bar(country_counts, x="Country", y="Companies", title="Companies by Country")
fig.show()
```

### 12. Interactive histogram — annual-revenue distribution

```python
fig = px.histogram(df, x="Annual_Revenue", nbins=15, title="Annual Revenue Distribution")
fig.show()
```

### 13. Interactive scatter plot — market share versus profit margin

```python
fig = px.scatter(df, x="Market_Share", y="Profit_Margin", title="Market Share vs Profit Margin")
fig.show()
```

### 14. Interactive box plot — salary by ownership type

```python
fig = px.box(df, x="Ownership_Type", y="Average_Salary", title="Average Salary by Ownership Type")
fig.show()
```

### 15. Interactive pie chart — primary-cloud distribution

```python
cloud_counts = df["Primary_Cloud"].value_counts().reset_index()
cloud_counts.columns = ["Primary Cloud", "Companies"]
fig = px.pie(cloud_counts, names="Primary Cloud", values="Companies", title="Primary Cloud Distribution")
fig.show()
```
