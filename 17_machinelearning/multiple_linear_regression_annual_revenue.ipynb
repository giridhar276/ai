# Multiple Linear Regression – Annual Revenue Prediction

This notebook builds a simple **multiple linear regression** model to predict `Annual_Revenue` using a few numerical company features.

## 1. Import the required libraries

```python
import pandas as pd
import matplotlib.pyplot as plt
from sklearn.model_selection import train_test_split
from sklearn.linear_model import LinearRegression
from sklearn.metrics import mean_absolute_error, mean_squared_error, r2_score
```

## 2. Load the CSV file
Keep the CSV file in the same folder as this notebook.

```python
file_name = "software_companies_dataset_v2.csv"
df = pd.read_csv(file_name)
df.head()
```

## 3. Inspect the dataset

```python
print("Dataset shape:", df.shape)
print("\nMissing values in selected columns:")
selected_columns = ["Employees", "Market_Share", "Profit_Margin", "Adoption_Rate_AI", "Adoption_Rate_Cloud", "Annual_Revenue"]
display(df[selected_columns].isnull().sum())
df[selected_columns].describe()
```

## 4. Select input features and target

The target is `Annual_Revenue`. The five input features describe company size, market position, profitability, and technology adoption.

```python
features = ["Employees", "Market_Share", "Profit_Margin", "Adoption_Rate_AI", "Adoption_Rate_Cloud"]
target = "Annual_Revenue"

model_data = df[features + [target]].dropna()
X = model_data[features]
y = model_data[target]

print("Rows used for modelling:", len(model_data))
display(X.head())
display(y.head())
```

## 5. Split the data
We use 80% of the data for training and 20% for testing.

```python
X_train, X_test, y_train, y_test = train_test_split(
    X, y, test_size=0.20, random_state=42
)

print("Training rows:", len(X_train))
print("Testing rows:", len(X_test))
```

## 6. Train the multiple linear regression model

```python
model = LinearRegression()
model.fit(X_train, y_train)
print("Model training completed.")
```

## 7. Make predictions

```python
y_pred = model.predict(X_test)

results = pd.DataFrame({
    "Actual Revenue": y_test.values,
    "Predicted Revenue": y_pred
})
results.head(10)
```

## 8. Evaluate the model
Lower MAE and RMSE are better. An R² value closer to 1 indicates a better fit.

```python
mae = mean_absolute_error(y_test, y_pred)
mse = mean_squared_error(y_test, y_pred)
rmse = mse ** 0.5
r2 = r2_score(y_test, y_pred)

print(f"Mean Absolute Error (MAE): {mae:.2f}")
print(f"Root Mean Squared Error (RMSE): {rmse:.2f}")
print(f"R-squared (R²): {r2:.3f}")
```

## 9. View feature coefficients
A positive coefficient increases predicted revenue; a negative coefficient decreases it, while other features remain constant.

```python
coefficients = pd.DataFrame({
    "Feature": features,
    "Coefficient": model.coef_
}).sort_values("Coefficient", ascending=False)

print("Intercept:", model.intercept_)
coefficients
```

## 10. Plot actual versus predicted revenue

```python
plt.figure(figsize=(7, 5))
plt.scatter(y_test, y_pred, color="royalblue", alpha=0.7)
minimum = min(y_test.min(), y_pred.min())
maximum = max(y_test.max(), y_pred.max())
plt.plot([minimum, maximum], [minimum, maximum], color="red", linestyle="--")
plt.xlabel("Actual Annual Revenue")
plt.ylabel("Predicted Annual Revenue")
plt.title("Actual vs Predicted Annual Revenue")
plt.show()
```
